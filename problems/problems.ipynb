```python
import numpy as np
import pandas as pd

df = pd.read_csv('./data/ames_prices.csv')
df['central_air'] = (df['Central.Air'] == 'Y').astype(int)   # missing/other values -> 0

x = df['central_air'].to_numpy()
n = len(x)
```

```python
#7a

p_hat = x.mean()
print(f"(a) n = {n},  estimated proportion with central air p_hat = {p_hat:.4f}\n")
```

```python
#7b
seed = 42
B = 5000
rng = np.random.default_rng(seed)
boot_estimates = np.array([rng.choice(x, size=n, replace=True).mean() for _ in range(B)])
print(f"(b) seed = {seed},  bootstrap replications B = {B}")
```

```python
#7c
se_boot = boot_estimates.std(ddof=1)
print(f"(c) bootstrap SE = {se_boot:.5f}\n")
```

```python
#7d
p0 = 0.90
z_stat = (p_hat - p0) / se_boot
crit = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}

rows = []
for level, z in crit.items():
    lo, hi = p_hat - z * se_boot, p_hat + z * se_boot
    contains = lo <= p0 <= hi
    rows.append({
        'Confidence': f"{int(level*100)}%",
        'alpha': round(1 - level, 2),
        'Critical z': z,
        'Lower': round(lo, 4),
        'Upper': round(hi, 4),
        'Contains 0.90?': contains,
        'Decision': 'Fail to reject H0' if contains else 'Reject H0',
    })

results = pd.DataFrame(rows)
print("(d)/(e) Normal confidence intervals and tests of H0: p = 0.90")
print(f"      z = (p_hat - 0.90) / SE_boot = {z_stat:.2f}")
display(results)
```

7e. At alpha = .1 we reject the null hypothesis. At alpha = .05 we reject the hypthesis. At alpha = .001 we reject the hypothesis. .9 does not appear in any of the confidence intervals. This means that p does not equal .9.

```python
#8a
df = pd.read_csv('./data/ames_prices.csv')
d = df[['area', 'price']].dropna()          
x = np.arcsinh(d['area'].to_numpy())
y = np.arcsinh(d['price'].to_numpy())
n = len(d)

r_hat = np.corrcoef(x, y)[0, 1]
print(f"n = {n},  Pearson r(arcsinh(area), arcsinh(price)) = {r_hat:.4f}\n")

seed = 42
B = 5000
rng = np.random.default_rng(seed)
boot_r = np.empty(B)
for b in range(B):
    idx = rng.integers(0, n, size=n)        
    boot_r[b] = np.corrcoef(x[idx], y[idx])[0, 1]
print(f"(a) seed = {seed},  bootstrap replications B = {B}")
```

```python
#8b
se_boot = boot_r.std(ddof=1)
print(f"(b) bootstrap SE = {se_boot:.5f}\n")
```

```python
#8c & 8d
rho0 = 0.0
z_stat = (r_hat - rho0) / se_boot
crit = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}

rows = []
for level, z in crit.items():
    lo, hi = r_hat - z * se_boot, r_hat + z * se_boot
    contains = lo <= rho0 <= hi
    rows.append({
        'Confidence': f"{int(level*100)}%",
        'alpha': round(1 - level, 2),
        'Critical z': z,
        'Lower': round(lo, 4),
        'Upper': round(hi, 4),
        'Contains 0?': contains,
        'Decision': 'Fail to reject H0' if contains else 'Reject H0',
    })

results = pd.DataFrame(rows)
print("(c)/(d) Centered-Normal confidence intervals and tests of H0: rho = 0")
print(f"      z = r_hat / SE_boot = {z_stat:.2f}")
display(results)
```

8d. For alpha = .1 we reject the null hypthosis. At alpha = .05 we reject the hypthesis. At alpha = .001 we reject the hypothesis. 0 does npt appear in any of the confidence intervals. These results show a trend where larger houses sell for higher prices.

8e. The pairs must stay together for resampling because we are looking at the correlations. If they were separated we would be unable to tell if the area affected price.

```python
#9a
df = pd.read_csv('./data/metabric.csv')
os_months = df['Overall Survival (Months)'].dropna().to_numpy() 
n = len(os_months)
print(f"Rows in file = {len(df)},  non-missing survival times n = {n}")

mean_hat = os_months.mean()
print(f"Mean recorded time to death or last follow-up = {mean_hat:.2f} months "
      f"({mean_hat/12:.2f} years)\n")

seed = 42
B = 5000
rng = np.random.default_rng(seed)
boot_means = np.array([rng.choice(os_months, size=n, replace=True).mean() for _ in range(B)])
print(f"(a) seed = {seed},  bootstrap replications B = {B}")
```

```python
#9b
se_boot = boot_means.std(ddof=1)
print(f"(b) bootstrap SE = {se_boot:.3f} months ({se_boot/12:.3f} years)\n")
```

```python
#9c & d
mu0 = 120
z_stat = (mean_hat - mu0) / se_boot
crit = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}

rows = []
for level, z in crit.items():
    lo, hi = mean_hat - z * se_boot, mean_hat + z * se_boot
    contains = lo <= mu0 <= hi
    rows.append({
        'Confidence': f"{int(level*100)}%",
        'alpha': round(1 - level, 2),
        'Critical z': z,
        'Lower (mo)': round(lo, 2),
        'Upper (mo)': round(hi, 2),
        'Lower (yr)': round(lo / 12, 2),
        'Upper (yr)': round(hi / 12, 2),
        'Contains 120?': contains,
        'Decision': 'Fail to reject H0' if contains else 'Reject H0',
    })

results = pd.DataFrame(rows)
print("(c)/(d) Centered-Normal confidence intervals and tests of H0: mu = 120 months (10 years)")
print(f"      z = (mean_hat - 120) / SE_boot = {z_stat:.2f}")
display(results)
```

9d. The hypothesis is rejected at all alpha levels.

9e. The 95% confidence intercal is 10.41 years - 11.11 years. This means this is the estimate for the population mean for overall survival ending at death or last follow up. Since the study is still ongoing some patients have more years to live and are so their recorded result is lower than reality. The true mean would be larger for unrestricted lifetime.

```python
#10a
df = pd.read_csv('./data/nhanes_data_17_18.csv')

AGE_COL = 'AgeInYearsAtScreening'       
SLEEP_COL = 'SleepHoursWeekdaysOrWorkdays'
if AGE_COL not in df.columns:
    raise KeyError(f"'{AGE_COL}' not found. Age-like columns: "
                   f"{[c for c in df.columns if 'age' in c.lower()]}")

adults = df[(df[AGE_COL] >= 18) & (df[SLEEP_COL].notna())]
sleep = adults[SLEEP_COL].to_numpy()
n = len(sleep)


mean_hat = sleep.mean()
print(f"Adults (18+) with observed weekday sleep: n = {n}")
print(f"Mean weekday sleep = {mean_hat:.3f} hours ({mean_hat*60:.1f} minutes)\n")

seed = 20261023
B = 5000
rng = np.random.default_rng(seed)
boot_means = np.array([rng.choice(sleep, size=n, replace=True).mean() for _ in range(B)])
print(f"(a) seed = {seed},  bootstrap replications B = {B}")
```

```python
#10b
se_boot = boot_means.std(ddof=1)
print(f"(b) bootstrap SE = {se_boot:.4f} hours ({se_boot*60:.2f} minutes)\n")
```

```python
#10c & d
mu0 = 8
z_stat = (mean_hat - mu0) / se_boot
crit = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}

rows = []
for level, z in crit.items():
    lo, hi = mean_hat - z * se_boot, mean_hat + z * se_boot
    contains = lo <= mu0 <= hi
    rows.append({
        'Confidence': f"{int(level*100)}%",
        'alpha': round(1 - level, 2),
        'Critical z': z,
        'Lower (hr)': round(lo, 3),
        'Upper (hr)': round(hi, 3),
        'Contains 8?': contains,
        'Decision': 'Fail to reject H0' if contains else 'Reject H0',
    })

results = pd.DataFrame(rows)
print("(c)/(d) Centered-Normal confidence intervals and tests of H0: mu = 8 hours")
print(f"      z = (mean_hat - 8) / SE_boot = {z_stat:.2f}")
print(f"      Estimated difference from 8 hours = {(mean_hat - mu0)*60:+.1f} minutes")
display(results)
```

10d. The hypothesis is rejected at all values of alpha tested.

10e. The tests show that the mean is significantly different from 8, but when looking at the numbers practically the difference is around 20-30 minutes. This does not seem truly different from just looking at the numbers but the test shows it is statistically significant.
